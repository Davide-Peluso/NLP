```python
import os
import pandas as pd

# Usa la directory corrente (dove stai eseguendo il notebook)
folder = os.getcwd()

# Prendi tutti i file CSV nella cartella
csv_files = [f for f in os.listdir(folder) if f.endswith('.csv')]

# Leggi e accumula tutti i DataFrame
df_list = [pd.read_csv(os.path.join(folder, file)) for file in csv_files]

# Unisci i DataFrame
merged_df = pd.concat(df_list, ignore_index=True)

# Salva il CSV unificato
merged_df.to_csv(os.path.join(folder, 'merged.csv'), index=False)

print(f"Merged {len(csv_files)} CSV files into 'merged.csv'")
```
