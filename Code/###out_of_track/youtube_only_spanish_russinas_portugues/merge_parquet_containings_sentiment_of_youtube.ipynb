```python
import os
import pandas as pd
from pathlib import Path

# Cartella corrente (dove si trova lo script)
folder = Path.cwd()

# Trova tutti i file .parquet nella cartella
parquet_files = [f for f in folder.glob("*.parquet")]

# Leggi tutti i file parquet e uniscili
df = pd.concat((pd.read_parquet(f) for f in parquet_files), ignore_index=True)

print(f"{len(parquet_files)} file uniti, dataframe finale ha {len(df)} righe.")
```

```python
df.to_csv('df_youtube_sentiment_srp.csv', index=False)
```
